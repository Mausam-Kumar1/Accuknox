```python
import requests
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
url = "https://dummyjson.com/users?limit=10"
```

```python
response = requests.get(url)
```

```python
data = response.json()
```

```python
students_scores = []
```

```python
for user in data.get('users', []):
    name = user.get('firstName')
    score = round(user.get('weight', 75)) 
    students_scores.append({
        "student_name": name,
        "test_score": score
    })
```

```python
print("Fetched JSON Data from API:")
print(students_scores)
```

```python
df = pd.DataFrame(students_scores)
```

```python
average_score = df['test_score'].mean()
```

```python
print(" Student Scores Table ")
print(df)
print(f"\nAverage Test Score of Class: {average_score:.2f}")
```

```python
plt.figure(figsize = (10, 5))
```

```python
plt.bar(df['student_name'], df['test_score'], color = 'royalblue', edgecolor = 'black', width = 0.5)
plt.xticks(rotation=45)
plt.axhline(average_score, color='red', linestyle='--', label=f'Average Score ({average_score:.2f})')
plt.title('Student Test Scores Visualization', fontsize=14)
plt.xlabel('Student Name', fontsize=12)
plt.ylabel('Test Score', fontsize=12)
plt.ylim(0, 100)
plt.legend()
plt.show()
```
