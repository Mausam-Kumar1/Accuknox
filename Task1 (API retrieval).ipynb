```python
import requests
```

```python
import sqlite3
```

```python
url = "https://openlibrary.org/search.json?q=Artificial+intelligence&limit=10"
```

```python
response = requests.get(url)
```

```python
books = response.json()
```

```python
print("Status Code:", response.status_code)
print("Total books fetched in this batch:", len(books.get('docs', [])))
```

```python
conn = sqlite3.connect('AI books store.db')
cursor = conn.cursor()
```

```python
cursor.execute('''
    CREATE TABLE IF NOT EXISTS ai_books (
        book_id INTEGER PRIMARY KEY AUTOINCREMENT,
        title TEXT,
        author TEXT,
        publication_year TEXT
    )
''')
```

```python
cursor.execute('DELETE FROM ai_books')
```

```python
list = books.get('docs', [])
```

```python
for book in list:
    title = book.get('title', 'N/A')
    
    authors_list = book.get('author_name', ['Unknown Author'])
    author = ", ".join(authors_list)
    
    pub_year = str(book.get('first_publish_year', 'N/A'))
    
    cursor.execute('''
        INSERT INTO ai_books (title, author, publication_year)
        VALUES (?, ?, ?)
    ''', (title, author, pub_year))

print("Books Inserted!")
```

```python
conn.commit()
```

```python
cursor.execute("SELECT book_id, title, author, publication_year FROM ai_books")
```

```python
saved_books = cursor.fetchall()
```

```python
print(f"Total {len(saved_books)} books successfully retrieved from database")
```

```python
print(" AI BOOKS STORED IN SQLITE DATABASE \n")

for row in saved_books:
    print(f"Book ID : {row[0]}")
    print(f"Title   : {row[1]}")
    print(f"Author  : {row[2]}")
    print(f"Year    : {row[3]}")
    print("-" * 45)

conn.close()
print("\nDatabase connection closed.")
```
