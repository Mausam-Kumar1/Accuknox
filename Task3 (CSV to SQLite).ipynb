```python
import csv
```

```python
import sqlite3
```

```python
users = [
    ["name", "email"],
    ["Aarav Sharma", "aaravsharma@gmail.com"],
    ["Priya Patel", "priyapatel@gmail.com"],
    ["Rohan Verma", "rohanverma@gmail.com"],
    ["Sneha Gupta", "snehagupta@gmail.com"],
    ["Vikram Singh", "vikramsingh@gmail.com"]
]
```

```python
with open('users.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(users)

print("users.csv file created!")
```

```python
conn = sqlite3.connect('User data.db')
cursor = conn.cursor()
```

```python
cursor.execute('''
    CREATE TABLE IF NOT EXISTS Users (
        user_id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        email TEXT
    )
''')
```

```python
cursor.execute('DELETE FROM Users')
```

```python
with open('users.csv', mode='r') as file:
    csv_reader = csv.reader(file)
    header = next(csv_reader)
    for row in csv_reader:
        user_name = row[0]
        user_email = row[1]
        cursor.execute('''
            INSERT INTO Users (name, email)
            VALUES (?, ?)
        ''', (user_name, user_email))
```

```python
conn.commit()
print("CSV file data inserted in SQLite Database!")
```

```python
print(" USERS STORED IN SQLITE DATABASE \n")
cursor.execute("SELECT user_id, name, email FROM Users")
saved_users = cursor.fetchall()
for user in saved_users:
    print(f"ID: {user[0]} | Name: {user[1]} | Email: {user[2]}")
conn.close()
```
